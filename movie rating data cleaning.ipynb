# Movie Rating - Data Cleaning & Preprocessing
**Team 7** | Movie Rating & Audience Preference Analysis

Steps: Load raw data, handle missing values, remove duplicates, feature engineering, export.

```python
import pandas as pd
import numpy as np

df = pd.read_excel('raw_movie_ratings_data.xlsx')
print('Raw shape:', df.shape)
print('Missing values:\n', df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print('After dedup:', df.shape)
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df = df.dropna(subset=['rating'])
print('After null drop:', df.shape)
```

```python
df['rating_category'] = pd.cut(df['rating'],bins=[0,2,3,4,5],
    labels=['Poor','Average','Good','Excellent'])
df['liked'] = (df['rating'] >= 3.5).astype(int)
print(df[['rating','rating_category','liked']].head(10))
```

```python
df.to_excel('cleaned_movie_ratings_data.xlsx', index=False)
print('Exported! Shape:', df.shape)
```
